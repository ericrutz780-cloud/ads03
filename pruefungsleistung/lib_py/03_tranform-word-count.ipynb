## Integration von OpenAI API

### Titel: Inhalt kategorisieren

```python
title_df = cleaned_masterfile_df[["id", "title"]]
```

```python
from openai import OpenAI
from pydantic import BaseModel

API_KEY = ""
client = OpenAI(api_key=API_KEY)


output = []

for i in range(len(title_df["title"])):
    int_text = str(title_df.at[i, "title"])
    print(f"Analysiere Titel {i+1}/{len(title_df)}: {int_text[:30]}...")
    response = client.chat.completions.create(
        model = 'gpt-5.2',
        messages = [
          {'role': 'system', 'content': (
                "Du bist ein wissenschaftlicher Assistent für Medienanalyse. "
                "Klassifiziere Medientitel in 'Vertrauen' oder 'Skepsis' basierend auf folgenden Kriterien:\n"
                "VERTRAUEN: Fokus auf Durchbruch, historische Chancen, Legitimität der Akteure und Zukunftslösungen.\n"
                "SKEPSIS: Fokus auf ungelöste Probleme, Misstrauen gegenüber Motiven, fragilen Charakter oder PR-Manöver.\n"
                "Antworte nur mit einem der beiden Begriffe.")},
          {'role': 'user', 'content': f"Titel zur Analyse: {int_text}"}
        ]
      )      
    chat_response = response.choices[0].message.content
    output.append(chat_response)

title_df["analyse_title"] = output
```
